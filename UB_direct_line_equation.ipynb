```python
#Direct Line Equation Algorithm
import matplotlib.pyplot as plt

def get_line_points(x1, y1, x2, y2):
    pixels = []
    print("Using Direct Line Equation\n")
    print("Starting Point:", (x1, y1))
    print("Ending Point:", (x2, y2))
    dx = x2 - x1
    dy = y2 - y1
    # For only vertical line
    if dx == 0:
        print("Case: Vertical Line.\n")
        print("Slope is Undefined.\n")
        start_y = min(y1, y2)
        end_y = max(y1, y2)
        for y in range(start_y, end_y + 1):
            pixels.append((x1, y))

    else:
        m = dy / dx
        b = y1 - m * x1
        print("Slope :", m)
        print("Y-intercept :", b)

        if abs(m) <= 1:
            print("Case: Shallow Line.\n")
            start_x = min(x1, x2)
            end_x = max(x1, x2)
            for x in range(start_x, end_x + 1):
                y = (m * x + b)
                round_y = round(y)
                pixels.append((x, round_y))

                print(
                    "x=", x,
                    "y=", round_y,
                    "pixels=", (x, round_y)
                )

        else:
            print("Case: Steep Line.\n")
            start_y = min(y1, y2)
            end_y = max(y1, y2)
            for y in range(start_y, end_y + 1):
                x = ((y - b) / m)
                round_x = round(x)
                pixels.append((round_x, y))

                print(
                    "x=", round_x,
                    "y=", y,
                    "pixels=", (round_x, y)
                )

    return pixels


```

```python
def plot_line(pixels, x1, y1, x2, y2):
    xs = [p[0] for p in pixels]
    ys = [p[1] for p in pixels]

    plt.figure(figsize=(7, 6))
    plt.plot(xs, ys, color="blue", linewidth=1, label="Line")
    plt.scatter(xs, ys, color="red", s=25, zorder=5, label="Computed points")
    plt.scatter([x1, x2], [y1, y2], color="green", s=80,
                marker="o", zorder=6, label="Start / End")

    plt.title("Line Plot using Direct Line Equation (y = mx + b)")
    plt.xlabel("X-axis")
    plt.ylabel("Y-axis")
    plt.axhline(0, color="gray", linewidth=0.5)
    plt.axvline(0, color="gray", linewidth=0.5)
    plt.grid(True, linestyle="--", alpha=0.6)
    plt.legend()
    plt.axis("equal")
    plt.savefig("line_plot.png", dpi=150)
    print("\nPlot saved as 'line_plot.png'")
    plt.show()


def main():
    print("Enter the starting point:")
    x1 = int(input("  x1 = "))
    y1 = int(input("  y1 = "))

    print("Enter the ending point:")
    x2 = int(input("  x2 = "))
    y2 = int(input("  y2 = "))

    pixels = get_line_points(x1, y1, x2, y2)
    plot_line(pixels, x1, y1, x2, y2)


if __name__ == "__main__":
    main()

```
